## 4. 新增固定列与计算列

1. Organism：所有行定义为Human。
2. Tissue：所有行定义为Duodenum。
3. Inflammation_Score = (IFNG + CXCL10 + STAT1) / 3。
4. Barrier_Score = (OCLN + TJP1 + CLDN4) / 3。
5. Epithelial_Marker_Score = (EPCAM + KRT18) / 2。
6. Score_Difference = Inflammation_Score - Barrier_Score。
7. 显示计算所需原始列及新增列，检查结果。


```python
# 1, 2
df['Organism'] = 'Human'
df['Tissue'] = 'Duodenum'

# 3, 4, 5, 6
df["Inflammation_Score"] = (df['IFNG'] + df['CXCL10'] + df['STAT1']) / 3
df['Barrier_Score'] = (df["OCLN"] + df["TJP1"] + df["CLDN4"]) / 3
df["Epithelial_Marker_Score"] = (df["EPCAM"] + df["KRT18"]) / 2
df["Score_Difference"] = (df["Inflammation_Score"] - df["Barrier_Score"])

print(df)
```
